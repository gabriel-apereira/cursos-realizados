# Definição de Esquemas e Validações de Dados
Estrutura Básica de um Esquema

Um esquema no Pandera é definido usando a classe DataFrameSchema e consiste em:

Colunas e seus tipos

Regras de validação

Configurações adicionais

# Tipos de Dados Suportados

```python
import numpy as np
import pandera.pandas as pa

# Tipos básicos
schema_basico = pa.DataFrameSchema({
    'inteiro': pa.Column(int),
    'decimal': pa.Column(float),
    'texto': pa.Column(str),
    'booleano': pa.Column(bool),
    'data': pa.Column(pa.DateTime)
})

# Tipos do Pandas
schema_pandas = pa.DataFrameSchema({
    'categoria': pa.Column(pa.Category),
    'timedelta': pa.Column(pa.Timedelta)
})
```

# Validações Básicas

### Validação de Tipo

```python
schema_tipo = pa.DataFrameSchema({
    'idade': pa.Column(int),  # deve ser inteiro
    'nome': pa.Column(str),   # deve ser string
    'salario': pa.Column(float)  # deve ser float
})
```

### Validação de Intervalo

```python
schema_intervalo = pa.DataFrameSchema({
    'idade': pa.Column(
        int,
        checks=[
            pa.Check.ge(0),    # maior ou igual a 0
            pa.Check.le(120)   # menor ou igual a 120
        ]
    ),
    'nota': pa.Column(
        float,
        checks=[
            pa.Check.ge(0),    # maior ou igual a 0
            pa.Check.le(10)    # menor ou igual a 10
        ]
    )
})
```

# Validações Avançadas

### Validação de Formato

```python
schema_formato = pa.DataFrameSchema({
    'email': pa.Column(
        str,
        checks=pa.Check.str_matches(r'^[^@]+@[^@]+\.[^@]+$')
    ),
    'cpf': pa.Column(
        str,
        checks=pa.Check.str_matches(r'^\d{3}\.\d{3}\.\d{3}-\d{2}$')
    )
})
```

### Validação de Valores Únicos

```python
schema_unicos = pa.DataFrameSchema({
    'id': pa.Column(
        int,
        checks=pa.Check(lambda s: s.is_unique()),  # valores devem ser únicos
        nullable=False
    )
})
```

### Validação de Valores em Lista

```python
schema_lista = pa.DataFrameSchema({
    'estado': pa.Column(
        str,
        checks=pa.Check.isin(['SP', 'RJ', 'MG', 'RS'])
    ),
    'status': pa.Column(
        str,
        checks=pa.Check.isin(['ativo', 'inativo', 'pendente'])
    )
})
```

# Validações Personalizadas

### Funções de Validação Customizadas

```python
def verificar_soma_colunas(df):
    return df['coluna1'] + df['coluna2'] == df['total']

schema_custom = pa.DataFrameSchema({
    'coluna1': pa.Column(float),
    'coluna2': pa.Column(float),
    'total': pa.Column(float)
}, checks=pa.Check(verificar_soma_colunas))
```

### Validação com Expressões Lambda

```python
schema_lambda = pa.DataFrameSchema({
    'idade': pa.Column(
        int,
        checks=pa.Check(lambda x: x > 0)
    ),
    'salario': pa.Column(
        float,
        checks=pa.Check(lambda x: x > 1000)
    )
})
```

# Validação de DataFrame Completo

```python
schema_completo = pa.DataFrameSchema({
    'id': pa.Column(
        int,
        checks=[
            pa.Check(lambda x: x.is_unique()),
            pa.Check.gt(0)
        ],
        nullable=False
    ),
    'nome': pa.Column(
        str,
        checks=pa.Check.str_length(min_value=3, max_value=100),
        nullable=False
    ),
    'idade': pa.Column(
        int,
        checks=[
            pa.Check.ge(0),
            pa.Check.le(120)
        ]
    ),
    'email': pa.Column(
        str,
        checks=pa.Check.str_matches(r'^[^@]+@[^@]+\.[^@]+$'),
        nullable=False
    ),
    'salario': pa.Column(
        float,
        checks=[
            pa.Check.gt(0),
            pa.Check.lt(1000000)
        ]
    )
})
```
