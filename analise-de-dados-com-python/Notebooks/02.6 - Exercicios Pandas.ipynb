#  Exercícios de Fixação

1 - Pessoa mais jovem por departamento:
Encontre, para cada departamento, o nome e a idade da pessoa mais jovem.

2 - Média de idade por cidade:
Calcule a idade média dos colaboradores, agrupando por cidade.

3 - Maior salário por cidade:
Identifique, para cada cidade, o maior salário registrado.

4 - Colaboradores acima da média salarial geral:
Filtre e liste todas as pessoas cujo salário está acima da média geral do DataFrame.

```python
import pandas as pd
import numpy as np

# Criando um DataFrame de exemplo
dados = {
    'Nome': ['João', 'Maria', 'Pedro', 'Ana', 'Carlos', 'Julia', 'Lucas', 'Beatriz'],
    'Idade': [25, 30, 35, 28, 40, 22, 33, 27],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'São Paulo', 
               'Rio de Janeiro', 'Belo Horizonte', 'São Paulo', 'Rio de Janeiro'],
    'Salário': [5000, 6000, 4500, 7000, 5500, 4000, 6500, 4800],
    'Departamento': ['TI', 'RH', 'Vendas', 'TI', 'Vendas', 'RH', 'TI', 'Vendas']
}

df = pd.DataFrame(dados)
```

### 1 - Pessoa mais jovem por departamento:
Encontre, para cada departamento, o nome e a idade da pessoa mais jovem.

```python
df_jovem = df.loc[df.groupby('Departamento')['Idade'].idxmin()]
df_jovem
```

### 2 - Média de idade por cidade:
Calcule a idade média dos colaboradores, agrupando por cidade.

```python
df_cidade = df.groupby('Cidade')['Idade'].mean()
df_cidade
```

### 3 - Maior salário por cidade:
Identifique, para cada cidade, o maior salário registrado.

```python
df_cidade_max = df.loc[df.groupby('Cidade')['Salário'].idxmax()]
df_cidade_max
```

### 4 - Colaboradores acima da média salarial geral:
Filtre e liste todas as pessoas cujo salário está acima da média geral do DataFrame.

```python
df_media = df[df['Salário'] > df['Salário'].mean()]
df_media
```
