# Introdução ao Seaborn e Matplotlib
Neste notebook, vamos explorar as bibliotecas de visualização de dados mais populares em Python: Matplotlib e Seaborn.

## Matplotlib
Matplotlib é a biblioteca fundamental para visualização de dados em Python. Ela fornece uma interface de baixo nível para criar gráficos e visualizações.

## Seaborn
Seaborn é uma biblioteca de visualização estatística construída sobre o Matplotlib. Ela fornece uma interface de alto nível para criar gráficos estatísticos atraentes e informativos.

```python
# Importando as bibliotecas necessárias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo do Seaborn
sns.set_theme()

# Configurando o tamanho padrão das figuras
plt.rcParams['figure.figsize'] = [10, 6]
```

## Comparação entre Matplotlib e Seaborn
Vamos criar um exemplo simples para mostrar a diferença entre Matplotlib e Seaborn:

```python
# Criando dados de exemplo
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Gráfico com Matplotlib
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(x, y)
plt.title('Gráfico com Matplotlib')
plt.xlabel('X')
plt.ylabel('Y')

# Gráfico com Seaborn
plt.subplot(1, 2, 2)
sns.lineplot(x=x, y=y)
plt.title('Gráfico com Seaborn')
plt.xlabel('X')
plt.ylabel('Y')

plt.tight_layout()
plt.show()
```

# Características principais do Seaborn
Estilo visual aprimorado: Seaborn vem com temas e paletas de cores pré-definidas
    
Visualizações estatísticas: Funções específicas para visualização de dados estatísticos
    
Integração com Pandas: Trabalha perfeitamente com DataFrames
    
Facilidade de uso: API mais simples e intuitiva
