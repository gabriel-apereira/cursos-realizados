# Filtragem, Seleção e Agregação de Dados
Este notebook demonstra técnicas fundamentais de manipulação de dados em Python usando pandas:

Filtragem de dados

Seleção de colunas e linhas

Agregação de dados

```python
import pandas as pd
import numpy as np

# Criando um DataFrame de exemplo
dados = {
    'Nome': ['João', 'Maria', 'Pedro', 'Ana', 'Carlos', 'Julia', 'Lucas', 'Beatriz'],
    'Idade': [25, 30, 35, 28, 40, 22, 33, 27],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'São Paulo', 
               'Rio de Janeiro', 'Belo Horizonte', 'São Paulo', 'Rio de Janeiro'],
    'Salário': [5000, 6000, 4500, 7000, 5500, 4000, 6500, 4800],
    'Departamento': ['TI', 'RH', 'Vendas', 'TI', 'Vendas', 'RH', 'TI', 'Vendas']
}

df = pd.DataFrame(dados)
df
```

## 1. Filtragem de Dados
A filtragem permite selecionar linhas com base em condições específicas.

```python
df['Salário'] > 50009
```

```python
# Filtrando pessoas com salário maior que 5000
df[df['Salário'] > 5000]
```

```python
# Filtrando pessoas do departamento de TI
df[df['Departamento'] == 'TI']
```

```python
# Filtrando com múltiplas condições (AND)
df[(df['Salário'] > 5000) & (df['Idade'] < 30)]
```

## 2. Seleção de Dados
A seleção permite escolher colunas específicas ou combinações de linhas e colunas.

```python
df[['Nome', 'Salário']]
```

```python
# Selecionando colunas específicas
df[['Nome', 'Salário', 'Departamento']]
```

```python
# Selecionando linhas e colunas específicas
df.loc[df['Idade'] > 30, 'Salário']
```

## 3. Agregação de Dados
A agregação permite resumir os dados usando funções estatísticas.

```python
# Média de salário por departamento
df.groupby('Departamento')['Salário'].mean()
```

```python
# Múltiplas agregações por departamento
df.groupby('Departamento').agg({
    'Salário': ['mean', 'min', 'max', 'count'],
    'Idade': 'mean'
})
```

```python
# Agregação por múltiplas colunas
df.groupby(['Departamento', 'Cidade'])['Salário'].mean()
```

## 4. Exemplo Prático Combinado
Vamos combinar filtragem, seleção e agregação em um exemplo prático.

```python
# Exemplo: Média salarial por departamento para pessoas com mais de 25 anos
df[df['Idade'] > 25].groupby('Departamento')['Salário'].mean()
```
