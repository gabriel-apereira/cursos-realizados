# Gráficos Avançados com Seaborn

Neste notebook, vamos explorar os tipos avançados de gráficos disponíveis no Seaborn:
- Heatmap
- Pairplot
- Violinplot
- Jointplot
- FacetGrid

```python
# Importando as bibliotecas necessárias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo do Seaborn
sns.set_theme()

# Configurando o tamanho padrão das figuras
plt.rcParams['figure.figsize'] = [10, 6]
```

## 1. Heatmap

Heatmaps são úteis para visualizar matrizes de dados, como matrizes de correlação.

```python
# Criando dados de exemplo
np.random.seed(42)
dados = np.random.randn(10, 10)
df = pd.DataFrame(dados, columns=[f'Var{i+1}' for i in range(10)])

# Calculando a matriz de correlação
corr = df.corr()

# Criando o heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title('Matriz de Correlação')
plt.show()
```

## 2. Pairplot

Pairplot cria uma matriz de gráficos de dispersão para visualizar relações entre múltiplas variáveis.

```python
# Criando dados de exemplo
np.random.seed(42)
df = pd.DataFrame({
    'x': np.random.normal(0, 1, 100),
    'y': np.random.normal(0, 1, 100),
    'z': np.random.normal(0, 1, 100),
    'categoria': np.random.choice(['A', 'B'], 100)
})

# Criando o pairplot
sns.pairplot(df, hue='categoria', diag_kind='kde')
plt.show()
```

## 3. Violinplot

Violinplot combina um boxplot com uma estimativa de densidade de kernel.

```python
# Criando dados de exemplo
np.random.seed(42)
df = pd.DataFrame({
    'valor': np.concatenate([
        np.random.normal(0, 1, 100),
        np.random.normal(2, 1, 100),
        np.random.normal(-2, 1, 100)
    ]),
    'grupo': ['A'] * 100 + ['B'] * 100 + ['C'] * 100
})

# Criando o violinplot
plt.figure(figsize=(10, 6))
sns.violinplot(data=df, x='grupo', y='valor')
plt.title('Distribuição dos Valores por Grupo')
plt.show()
```

## 4. Jointplot

Jointplot combina diferentes tipos de gráficos para visualizar a relação entre duas variáveis.

```python
# Criando dados de exemplo
np.random.seed(42)
x = np.random.normal(0, 1, 100)
y = 2 * x + np.random.normal(0, 0.5, 100)

# Criando o jointplot
sns.jointplot(x=x, y=y, kind='reg')
plt.show()
```

## 5. FacetGrid

FacetGrid permite criar múltiplos gráficos baseados em diferentes categorias dos dados.

```python
# Criando dados de exemplo
np.random.seed(42)
df = pd.DataFrame({
    'x': np.random.normal(0, 1, 200),
    'y': np.random.normal(0, 1, 200),
    'categoria1': np.random.choice(['A', 'B'], 200),
    'categoria2': np.random.choice(['X', 'Y'], 200)
})

# Criando o FacetGrid
g = sns.FacetGrid(df, col='categoria1', row='categoria2')
g.map_dataframe(sns.scatterplot, x='x', y='y')
g.add_legend()
plt.show()
```
