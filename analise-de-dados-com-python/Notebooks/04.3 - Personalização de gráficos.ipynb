# Personalização de Gráficos com Seaborn
Neste notebook, vamos explorar as diferentes formas de personalizar gráficos no Seaborn:

Cores e paletas

Estilos e temas

Legendas e rótulos

Tamanhos e layouts

```python
# Importando as bibliotecas necessárias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo do Seaborn
sns.set_theme()

# Configurando o tamanho padrão das figuras
plt.rcParams['figure.figsize'] = [10, 6]
```

### 1. Cores e Paletas
O Seaborn oferece várias paletas de cores predefinidas e permite criar paletas personalizadas.

```python
# Visualizando paletas de cores disponíveis
paletas = ['deep', 'muted', 'pastel', 'bright', 'dark', 'colorblind']
plt.figure(figsize=(15, 2))
for i, paleta in enumerate(paletas):
    plt.subplot(1, len(paletas), i+1)
    sns.palplot(sns.color_palette(paleta))
    plt.title(paleta)
plt.tight_layout()
plt.show()
```

### 2. Estilos e Temas
O Seaborn oferece diferentes estilos e temas para personalizar a aparência dos gráficos.

```python
# Criando dados de exemplo
x = np.linspace(0, 10, 100)
y = np.sin(x)

# Demonstrando diferentes estilos
estilos = ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks']
plt.figure(figsize=(15, 10))

for i, estilo in enumerate(estilos):
    plt.subplot(3, 2, i+1)
    sns.set_style(estilo)
    sns.lineplot(x=x, y=y)
    plt.title(f'Estilo: {estilo}')

plt.tight_layout()
plt.show()
```

### 3. Legendas e Rótulos
Personalizando legendas, títulos e rótulos dos eixos.

```python
# Criando dados de exemplo
np.random.seed(42)
df = pd.DataFrame({
    'x': np.random.normal(0, 1, 100),
    'y': np.random.normal(0, 1, 100),
    'categoria': np.random.choice(['A', 'B', 'C'], 100)
})

# Criando um gráfico personalizado
plt.figure(figsize=(10, 6))
g = sns.scatterplot(data=df, x='x', y='y', hue='categoria')

# Personalizando o gráfico
g.set_title('Gráfico de Dispersão Personalizado', fontsize=16, pad=20)
g.set_xlabel('Eixo X', fontsize=12)
g.set_ylabel('Eixo Y', fontsize=12)
g.legend(title='Categorias', title_fontsize=12, fontsize=10)

plt.show()
```

### 4. Tamanhos e Layouts
Ajustando o tamanho das figuras e organizando múltiplos gráficos.

```python
# Criando uma figura com múltiplos subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.suptitle('Múltiplos Gráficos', fontsize=16, y=1.02)

# Gráfico 1: Dispersão
sns.scatterplot(data=df, x='x', y='y', ax=axes[0,0])
axes[0,0].set_title('Dispersão')

# Gráfico 2: Histograma
sns.histplot(data=df, x='x', ax=axes[0,1])
axes[0,1].set_title('Histograma')

# Gráfico 3: Boxplot
sns.boxplot(data=df, x='categoria', y='y', ax=axes[1,0])
axes[1,0].set_title('Boxplot')

# Gráfico 4: Violinplot
sns.violinplot(data=df, x='categoria', y='y', ax=axes[1,1])
axes[1,1].set_title('Violinplot')

plt.tight_layout()
plt.show()
```
