## Carregamento e Exportação de Dados em Python

Neste notebook, vamos aprender como:

Carregar dados de arquivos CSV

Carregar dados de arquivos JSON

Exportar dados para CSV

Exportar dados para JSON

Primeiro, vamos importar as bibliotecas necessárias:

```python
import pandas as pd
import json
```

### 1 - Carregando csv

```python
# Carregando o arquivo CSV
df_vendas = pd.read_csv('vendas.csv')

# Exibindo as primeiras linhas
print("Primeiras 5 linhas do arquivo de vendas:")
display(df_vendas.head())

# Informações sobre o DataFrame
print("\nInformações sobre o DataFrame:")
display(df_vendas.info())
```

### 2 - Carregando JSON

```python
# Carregando o arquivo JSON
with open('cadastro.json', 'r', encoding='utf-8') as file:
    dados_json = json.load(file) # dict

print(type(dados_json))

# Convertendo para DataFrame
df_clientes = pd.DataFrame(dados_json['clientes'])

# Exibindo as primeiras linhas
print("Primeiras 5 linhas do arquivo de cadastro se houver:")
display(df_clientes.head())

# Informações sobre o DataFrame
print("\nInformações sobre o DataFrame:")
display(df_clientes.info())
```

```python
df_endereco = pd.DataFrame([dados_json['endereco']])
print(df_endereco)
```

### 3 - Exportando para csv

```python
# Criando um DataFrame com análises de vendas
vendas_por_cliente = df_vendas.groupby('cliente').agg({
    'quantidade': 'sum',
    'valor_unitario': lambda x: (x * df_vendas.loc[x.index, 'quantidade']).sum()
}).reset_index()

# Renomeando as colunas
vendas_por_cliente.columns = ['cliente', 'total_itens', 'valor_total']

# Exibindo o resultado
print("Análise de vendas por cliente:")
display(vendas_por_cliente)

# Exportando para CSV
vendas_por_cliente.to_csv('vendas_por_cliente.csv', index=False)
print("\nArquivo 'vendas_por_cliente.csv' foi criado com sucesso!")
```

### 4 - Exportando para csv

```python
# Combinando dados de vendas e clientes
vendas_detalhadas = df_vendas.merge(df_clientes, left_on='cliente', right_on='nome')

# Convertendo para dicionário
vendas_dict = {
    'vendas_detalhadas': vendas_detalhadas.to_dict('records')
}

# Exportando para JSON
with open('vendas_detalhadas.json', 'w', encoding='utf-8') as file:
    json.dump(vendas_dict, file, ensure_ascii=False, indent=4)

print("Arquivo 'vendas_detalhadas.json' foi criado com sucesso!")
```

# Resumo

Neste tutorial, aprendemos:

Como carregar dados de arquivos CSV usando pandas.read_csv()

Como carregar dados de arquivos JSON usando json.load()

Como exportar dados para CSV usando DataFrame.to_csv()
Como exportar dados para JSON usando json.dump()
Estas são operações fundamentais para trabalhar com dados em Python, especialmente em análise de dados e ciência de dados.
