# Gráficos Básicos com Seaborn
Neste notebook, vamos explorar os tipos básicos de gráficos disponíveis no Seaborn:

Gráficos de linha (lineplot)

Gráficos de barras (barplot)

Gráficos de dispersão (scatterplot)

Histogramas (histplot)

```python
# Importando as bibliotecas necessárias
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configurando o estilo do Seaborn
sns.set_theme()

# Configurando o tamanho padrão das figuras
plt.rcParams['figure.figsize'] = [10, 6]
```

### 1. Gráficos de Linha (lineplot)
Gráficos de linha são úteis para visualizar tendências ao longo do tempo ou para mostrar a relação entre variáveis contínuas.

```python
# Criando dados de exemplo
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Criando um DataFrame
df = pd.DataFrame({
    'x': np.concatenate([x, x]),
    'y': np.concatenate([y1, y2]),
    'função': ['seno'] * 100 + ['cosseno'] * 100
})

# Criando o gráfico de linha
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x='x', y='y', hue='função')
plt.title('Funções Seno e Cosseno')
plt.show()
```

### 2. Gráficos de Barras (barplot)
Gráficos de barras são ideais para comparar valores entre diferentes categorias.

```python
# Criando dados de exemplo
categorias = ['A', 'B', 'C', 'D', 'E']
valores = np.random.randint(10, 100, size=5)

# Criando o gráfico de barras
plt.figure(figsize=(10, 6))
sns.barplot(x=categorias, y=valores)
plt.title('Gráfico de Barras')
plt.xlabel('Categorias')
plt.ylabel('Valores')
plt.show()
```

### 3. Gráficos de Dispersão (scatterplot)
Gráficos de dispersão são úteis para visualizar a relação entre duas variáveis contínuas.

```python
# Criando dados de exemplo
dados = np.random.normal(0, 1, 1000)

# Criando o histograma
plt.figure(figsize=(10, 6))
sns.histplot(data=dados, bins=30, kde=True)
plt.title('Histograma com Curva de Densidade')
plt.xlabel('Valores')
plt.ylabel('Frequência')
plt.show()
```
