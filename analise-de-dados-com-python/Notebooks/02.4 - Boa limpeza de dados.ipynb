# 1. Remover ou Tratar Valores Nulos
Os valores nulos podem ser indicativos de dados ausentes ou podem conter informações importantes dependendo do contexto. As principais abordagens para lidar com esses valores incluem:

```python
import pandas as pd

df = pd.DataFrame({
    'Nome': ['Alice', 'Bob', 'Charlie', None],
    'Idade': [25, None, 30, 40],
    'Salario': [5000, 6000, None, 8000]
})

print(df.isnull().sum())  # Conta os valores nulos por coluna
```

```python
df_limpo = df.dropna()
print(df_limpo)
```

```python
df['Idade'].fillna(df['Idade'].median(), inplace=True)
df['Salario'].fillna(0, inplace=True)
```

# 2. Remover Duplicatas
Duplicatas podem distorcer análises estatísticas e previsões.

```python
df = df.drop_duplicates()
```

```python
df = df.drop_duplicates(subset=['Nome'], keep='first')
```

# 3. Tratar Valores Inconsistentes
Erros tipográficos ou formatação inconsistente podem comprometer a análise.

```python
df['Nome'] = df['Nome'].str.strip().str.title()
```

# 4. Tratar Outliers
Valores extremos podem distorcer a análise. Uma abordagem comum é remover outliers com base no intervalo interquartil:

```python
Q1 = df['Salario'].quantile(0.25)
Q3 = df['Salario'].quantile(0.75)
IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR
limite_superior = Q3 + 1.5 * IQR

df = df[(df['Salario'] >= limite_inferior) & (df['Salario'] <= limite_superior)]
```

# 5. Converter Tipos de Dados
Tipos de dados incorretos podem gerar erros em cálculos e consultas

```python
df['Idade'] = df['Idade'].astype(int)  # Convertendo para inteiro
```
