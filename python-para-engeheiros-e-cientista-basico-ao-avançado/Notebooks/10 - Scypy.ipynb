$$ A=
 \begin{bmatrix}

$$ X=

## 10.1.4 Exercício 3
Ache os autovalores e autovetores para as matrizes abaixo.

$$ A=
 \begin{bmatrix}
   -1 & 6 \\\\
   0 & 5
  \end{bmatrix}
$$

$$ B=
 \begin{bmatrix}
   -1/3 & 0 & 0 \\\\
   0 & 1 & 0 \\\\
   0 & 0 & 1/2
  \end{bmatrix}
$$

O autovalor e autovetor se associam à matriz da seguinte forma:

$$ [A] \{X\} = \lambda \{X\} $$

```python
import numpy as np
from scipy import linalg
```

```python
A = np.array([[-1,6],[0,5]])
```

```python
autovalor,autovetor = linalg.eig(A)
autovetor
```

```python
autovetor[::,0]
```

```python
A@autovetor[::,0]
```

## 10.2.4 - EDO

**from scipy.integrate import odeint**


| Funções | Descrição |
| :-- | :-- |
| integrate.odeint(func, y0, t) | Calcula a dada equação diferencial  |



https://docs.scipy.org/doc/scipy/reference/generated/scipy.integrate.odeint.html#scipy.integrate.odeint

```python
import sympy as sp

x = sp.symbols('x')

y_x = sp.Function('y_x')

dy_x = sp.Derivative(y_x(x),x)

eq = dy_x + 10*x

sp.dsolve(eq)
```

```python
from scipy.integrate import odeint
```

```python
def eq_dif(u,x):
    y,dy = u
    func = [dy, -10*x]
    return func
```

```python
x = list(range(11))
x
```

```python
odeint(eq_dif,[0,0],x)
```
