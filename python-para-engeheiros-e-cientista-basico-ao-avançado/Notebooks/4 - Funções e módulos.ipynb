## 4.3 Exercício 1 - 
Crie uma função que receba uma frase e retorne quantas palavras tem a frase. Dica: Utilize o método split().

```python
def qtd_palavras(Frase):
    quantidade = len(Frase.split(' '))
    return quantidade

qtd_palavras('Oi , como você está ?')
```

## 4.4 Exercício 2 -
Crie uma função que lê um arquivo de extensão .txt e retorne quantas palavras tem esse arquivo. Dica: utilize o método readlines()


```python
def contar_palavras_txt(caminho_arquivo):
    with open(caminho_arquivo, 'r') as arquivo:
        linhas = arquivo.readlines()
        
    total_palavras = 0
    for linha in linhas:
        palavras = linha.strip().split()
        total_palavras += len(palavras)
    
    return total_palavras

file_path = 'C:/Users/gabriel.pereira/OneDrive - Grupo ITG/Área de Trabalho/Cursos/Python Basico avanç Udemy/Python-para-Engenheiros-e-Cientistas-main/Notebooks/arquivos_auxiliares/PIB_100_maiores_cidades_2017.txt'

contar_palavras_txt(file_path)
```
