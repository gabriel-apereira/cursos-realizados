```python
for i in Test_numbers:
    file_name = path + str(i) + '.txt'
    print(file_name)
```

```python
df_list = []

for i in Test_numbers:
    file_name = path + str(i) + '.txt'
    df_list.append(pd.read_csv(file_name,skiprows = 6))
```

```python
concat_df = pd.concat(df_list, ignore_index = True)
```
