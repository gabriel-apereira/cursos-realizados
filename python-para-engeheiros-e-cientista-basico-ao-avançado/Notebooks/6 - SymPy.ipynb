## 6.1 Introdução ao Spyder a ao SymPy

Algumas de suas funcionalidades são:

SymPy é uma biblioteca para trabalhar com matemática simbólica. 

Algumas funcionalidades são:
- Trabalhar com matrizes

- Utilizar *solvers*
- Para outras funcionalidades acesse: https://docs.sympy.org/latest/index.html

## 6.10 Exercício 5
Dadas as matrizes, resolva o sistema de equação $[A].\{X\} = \{B\}$

$$ A=
 \begin{bmatrix}
   1 & 1 & 1 \\\\
   1 & 2 & 2 \\\\
   2 & 1 & 3 
\end{bmatrix}
$$

$$ X=
 \begin{bmatrix}
   x_1 \\\\
   x_2 \\\\
   x_3 
\end{bmatrix}
$$

$$ B=
 \begin{bmatrix}
   6 \\\\
   9 \\\\
   11 
\end{bmatrix}
$$




```python
import sympy as sp
sp.init_printing()

x1,x2,x3 = sp.symbols(['x1','x2','x3'])

X = sp.Matrix([[x1,x2,x3]]).T
A = sp.Matrix([[1,1,1],[1,2,2],[2,1,3]])
B = sp.Matrix([[6,9,11]]).T

#AX = B
# AX - B = 0

sistema = A*X - B

sp.solve(sistema,(x1,x2,x3))
```

## 6.11 Exercício 6
Resolva as EDOs:
$$ A)\ \ \frac{dy}{dx} = \frac{x^2}{y} $$

$$ B)\ \ m\ddot{x}(t) + c\dot{x}(t) + kx(t)=0; \ \ com\ x(0) = x_0 \ e\  \dot{x}(0) = v_0   $$




```python
import sympy as sp
sp.init_printing()
```

```python
#A

x= sp.symbols('x')
y_x = sp.Function('y_x')
dy_x = sp.Derivative(y_x(x),x)

edo1 = dy_x - x**2/(y_x(x))

sp.dsolve(edo1)
```

```python
#B

t,m,c,k,x0,v0 = sp.symbols(['t','m','c','k','x0','v0'])

x_t = sp.Function('x_t')
dx = sp.Derivative(x_t(t),t)
ddx = sp.Derivative(x_t(t),t,t)

edo2 = m*ddx + c*dx + k*x_t(t)

print(sp.dsolve(edo2))
```

```python
C1,C2 = sp.symbols(['C1','C2'])

x_t_solved = C1*sp.exp(t*(-c - sp.sqrt(c**2 - 4*k*m))/(2*m)) + C2*sp.exp(t*(-c + sp.sqrt(c**2 - 4*k*m))/(2*m))

#Condições iniciais

x_t_solved
dx_t_solved = sp.diff(x_t_solved,t)
dx_t_solved
```

```python
eq_1 = x_t_solved.subs(t,0) - x0
C1_solved = sp.solve(eq_1,C1)[0]

```

```python
x_t_solved = x_t_solved.subs(C1,C1_solved)
dx_t_solved = dx_t_solved.subs(C1,C1_solved)

eq_2 = dx_t_solved.subs(t,0) - v0

eq_2

C2_solved = sp.solve(eq_2,C2)
C2_solved
```

```python
x_t_solved = x_t_solved.subs(C2,C2_solved)
x_t_solved
```
